## 딕셔너리 기본과 조작
딕셔너리는 키-값 쌍으로 데이터를 저장하는 구조입니다.<br>
키를 통해 값을 효율적으로 조회하거나 수정할 수 있습니다.<br>

딕셔너리 `person = {'name': 'Alice', 'age': 25}`에서 이름(name)을 출력해보세요.

`in`을 활용해 키 `'age'`가 `person` 딕셔너리에 존재하는지 확인해보세요.

```python
person = {'name': 'Alice', 'age': 25}
print('age' in person)
```

`get()`을 활용해 키 `'name'`에 해당하는 값을 가져와 출력해보세요.

```python
person = {'name': 'Alice', 'age': 25}
print(person.get('name'))
```

`keys()`를 활용해 딕셔너리의 모든 키를 출력해보세요.

```python
person = {'name': 'Alice', 'age': 25}
print(person.keys())
```

`values()`를 활용해 딕셔너리의 모든 값을 출력해보세요.

```python
# 딕셔너리 'person'을 다시 생성합니다.
person = {'name': 'Alice', 'age': 25}
# 딕셔너리의 모든 '값'들을 출력합니다.
print(person.values())
```

`items()`를 활용해 딕셔너리의 모든 키-값 쌍을 출력해보세요.

```python
# 딕셔너리 'person'을 다시 생성합니다.
person = {'name': 'Alice', 'age': 25}
# 딕셔너리의 모든 '키-값' 쌍을 출력합니다.
print(person.items())
```

`del`을 활용해 키 `'age'`와 해당 값을 삭제해보세요.

```python
# 딕셔너리 'person'을 다시 생성합니다.
person = {'name': 'Alice', 'age': 25}
# 'age' 키와 해당하는 값을 삭제합니다.
del person['age']
# 삭제 후 딕셔너리를 출력하여 확인합니다.
print(person)
```

## 집합(Set) 사용법
집합은 중복을 허용하지 않는 비순서 컬렉션입니다.<br>
수학적 집합 연산이나 중복 제거에 유용하게 사용됩니다.<br>

집합 `s = set([1, 2, 2, 3])`을 생성하고 출력해보세요.

```python
# 중복된 요소를 가진 리스트로 집합 's'를 생성합니다.
s = set([1, 2, 2, 3])
# 생성된 집합을 출력합니다. 중복된 2는 하나만 남습니다.
print(s)
```

`add()`를 활용해 집합 `s`에 숫자 4를 추가해보세요.

```python
# 집합 's'를 다시 생성합니다.
s = set([1, 2, 3])
# 집합에 숫자 4를 추가합니다.
s.add(4)
# 변경된 집합을 출력합니다.
print(s)
```

`remove()`를 활용해 집합 `s`에서 숫자 2를 제거해보세요.

```python
# 집합 's'를 다시 생성합니다.
s = set([1, 2, 3])  # 집합 생성(중복 제거됨)
# 집합에서 숫자 2를 제거합니다.
s.remove(2)
# 변경된 집합을 출력합니다.
print(s)

```

`pop()`을 활용해 집합 `s`에서 임의의 요소를 제거하고 출력해보세요.

```python
# 집합 's'를 다시 생성합니다.
s = set([1, 3, 4])
# 집합에서 임의의 요소를 제거하고 그 요소를 출력합니다.
popped_element = s.pop()
print(f"제거된 요소: {popped_element}")
# 변경된 집합을 출력합니다.
print(s)
```

`in`을 활용해 숫자 1이 집합 `s`에 포함되어 있는지 확인해보세요.

```python
# 집합 's'를 다시 생성합니다.
s = set([1, 3]) # pop()으로 요소가 하나 제거되었다고 가정합니다.
# 숫자 1이 집합 's'에 포함되어 있는지 확인하고 결과를 출력합니다.
print(1 in s)
```
